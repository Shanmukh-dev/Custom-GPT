```python
from datasets import load_dataset
import json

dolly = load_dataset("databricks/databricks-dolly-15k")
# dolly = load_dataset("databricks/databricks-dolly-15k")

# print(dolly)


dolly_train = dolly["train"]

print(json.dumps(dolly_train[0], indent=2))
```

```python
oasst = load_dataset("OpenAssistant/oasst1")

oasst_train = oasst["train"]

print(json.dumps(oasst_train[0], indent=2))
```

```python
oasst_en = oasst_train.filter(
    lambda x: x["lang"] == "en"
)
oasst_en = oasst_en.filter(
    lambda x:(
        x["deleted"] is False
        and x["review_result"] is True
        and x["text"].strip() != ""
    )
)


print(oasst_en[0])
```

```python
print(json.dumps(oasst_en[0], indent = 2))
```

```python
def convert_dolly(sample):
    instruction = sample["instruction"].strip()
    context = sample["context"].strip()
    response = sample["response"].strip()

    instruction = instruction + ("\n\nContext:\n" + context if context else "")

    return {
        "messages":[
            {
                "role": "user",
                "content": instruction
            },
            {
                "role": "assistant",
                "content": response
            }
        ]
    }

dolly_converted = list(map(convert_dolly, dolly_train))

print(json.dumps(dolly_converted[0], indent=2))

with open("Data/dolly.json", "w") as f:

    json.dump(dolly_converted, f)
```

```python
text = ""
for sample in dolly_converted:
    msg = sample["messages"]

    query = f"<user>\n{msg[0]["content"]}\n</user>"
    response = f"<assistant>\n{msg[1]["content"]}\n</assistant>"

    text += f"{query}\n{response}\n<|endoftext|>\n\n"

with open("Data/dolly_train.txt", "w", encoding="utf-8") as f:
    f.write(text)
```

```python
oasst_messages = {}

for msg in oasst_en:
    oasst_messages[msg["message_id"]] = msg


def get_conversation(msg_id):
    conversation = []

    curr = msg_id
    while curr != None:
        msg = oasst_messages[curr]
        conversation.append({
            "role": msg["role"],
            "content": msg["text"]
        })

        curr = msg["parent_id"]

    conversation.reverse()

    return conversation
```

```python
def normalize_convo(convo):
    return [
        {
            "role": "user" if msg["role"] == "prompter" else "assistant",
            "content": msg["content"].strip()
        }
        for msg in convo
    ]
```
