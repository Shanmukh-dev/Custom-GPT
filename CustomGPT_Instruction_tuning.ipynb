# Preparing the Instruction Tuning dataset

```python
ds = datasets.load_dataset("PleIAS/SYNTH", split="train", streaming = True)
```

```python
config = GPTConfig(
    vocab_size = tokenizer.n_vocab,
    block_size = 256,

    d_model = 256,
    hidden_layers = 1024,
    n_heads = 4,
    n_layers = 6
)
```

```python
from tqdm.auto import tqdm
import textwrap
target_tokens = 250_000_000

tokens = []

pbar = tqdm(total=target_tokens)


for sample in ds:
    text = "" 
    if sample["language"] == "en":
        query = f"<User>\n{sample["query"]}\n</User>"
        response = textwrap.dedent(f"""
        <Assistant>
            <Reasoning>
            {sample["synthetic_reasoning"] if sample["synthetic_reasoning"] else ""}
            </Reasoning>

            <Assistant_Message>
            {sample["synthetic_answer"] if sample["synthetic_answer"] else ""}
            </Assistant_Message>
        </Assistant>
        """)
        text += query+"\n"+response

        ids = tokenizer.encode(text)
        tokens.extend(ids)

        currlen = min(len(tokens), target_tokens)

        pbar.n = currlen
        pbar.refresh()
        

    if len(tokens) >= target_tokens:
        break
pbar.close()
print("Dataset created")
```

```python
train_dataloader, test_dataloader = create_dataloaders(
    tokens, 
    0.9, 
    device, 
    config.block_size, 
    32
)
print(f"Train Dataloader length: {train_dataloader}")
print(f"Test Dataloader length: {test_dataloader}")
```

# Loading the model

```python
from torch.amp import GradScaler, autocast

modelV2 = CustomGPT(config)
state_dict = torch.load("custom_gpt_v2.pth", map_location=device)

state_dict = {k.removeprefix("module."): v for k, v in state_dict.items()}
modelV2.load_state_dict(state_dict)

if torch.cuda.device_count() > 1:
    modelV2 = nn.DataParallel(modelV2)
    print(f"Using {torch.cuda.device_count()} GPUs")

modelV2.to(device)


loss_fn = nn.CrossEntropyLoss()
optimizer = torch.optim.AdamW(modelV2.parameters(), lr=3e-4)
scaler = GradScaler()
```

```python
from tqdm.auto import tqdm

steps = 40000
train_iter = iter(train_dataloader)
torch.manual_seed(42)

for step in tqdm(range(1, steps+1)):
    # nonlocal train_iter
    modelV2.train()
    if step%100 == 0:
        print(f"Training step {step} started")
    
    train_loss, train_iter = train_step(
        modelV2,
        train_dataloader,
        train_iter,
        loss_fn,
        optimizer,
        scaler,
        device
    ) 
    if step%100 == 0:
        print(f"Training step {step} completed")

    if step % 5000 == 0:
        test_loss = test_step(modelV2, test_dataloader, 10, loss_fn, device)
        print(f"Step: {step} | Training loss: {train_loss} | Testing loss: {test_loss}")
```
